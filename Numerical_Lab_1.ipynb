## 5. Symplectic method for a purely oscillatory system ($\gamma = 0$)

When $\gamma = 0$, the Euler scheme has a structural problem: while the solutions are expected to be purely oscillatory (the eigenvalues of the associated matrix are purely imaginary), we find that they're spiraling out.
The problem would persist if we would switch to the implicit Euler scheme.

When $\gamma=0$, the system preserves the energy
$$
  E = \frac12(v^2 + \omega^2 x^2).
$$
Check it explicitly by computing $dE/dt$!


However, the Euler scheme increases the energy of the system: compute $E(t+h) - E(t)$! Can you prove that it increases exponentially with time?

A *symplectic* scheme  preserves the Hamiltonian structure of the dynamics. The fix looks innocent. Instead of
$$
  x_{n+1} = x_n + h v_n,
$$
$$
  v_{n+1} = v_n -h\omega^2 x_n,
$$
do sequencially
$$
  x_{n+1} = x_n + h v_n,
$$
$$
  v_{n+1} = v_n -h\omega^2 x_{n+1},
$$
i.e. update the velocity with the already updated value for the position.

This scheme has a clear physical origin: you keep on iterating two maps that have each an exact Hamiltonian structure. In the first step of a cycle, you update the position at a constant velocity. This is the motion generated by the Hamiltonian $H_1 = v^2/2$. In the second step of a cycle, you update the velocity whie keeping the position constant. This is the motion generated by the Hamiltonian $H_2 = \omega^2 x^2/2$.

## 6. Lotka-Volterra system

We could apply the explicit Euler scheme to solve the Lotka-Volterra system:
$$
  \frac{dx}{dt} = x(1-y),
  \qquad
  \frac{dy}{dt} = \epsilon y(x-1)
$$
for $\epsilon > 0$, and initial conditions
$$
  x(0) = x_0, \quad y(0) = y_0
$$
for some given $x_0,y_0$.


***Important caveat:*** The Lotka-Volterra system has an underlying symplectic structure. Because, the Euler scheme is not a symplectic scheme, numerical solutions will suffer from the same problem encountered above: solutions will diverge instead of forming closed orbits.


It is possible to design a symplectic scheme here as well, though it's less obvious to explain why it works, and we will simply give the solution here:
$$
  x_{n+1} = x_n e^{h(1-y_n)},
$$
$$
  y_{n+1} = y_n e^{h(x_{n+1} -1)}.
$$
